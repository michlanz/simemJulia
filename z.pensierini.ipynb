# Roadmap SIMEMJULIA

## 1. Modello Statico – Consolidamento

- Mantenere 200 ordini iniziali
- Fine simulazione a completamento naturale ordini
- Introdurre parametro orizzonte alternativo a 320 ore
- Salvare CSV completi
- Salvare CSV tagliati a 320 ore
- Implementare funzione di trim sui Dash prima del postprocess
- Rilanciare postprocessDF su Dash originali e tagliati
- Verificare coerenza tra completo e cut quando simulazione < 320
- Definire KPI baseline (throughput, completati, lateness, WIP, saturazione)

---

## 2. Separazione Simulazione / Analisi

- Mantenere simulazione indipendente da logiche di analisi
- Centralizzare trasformazioni in postprocessDF
- Evitare calcoli di policy dentro output
- Standardizzare naming e struttura CSV

---

## 3. Stabilizzazione Strutture Dati

- Revisione struttura Cliente
- Aggiungere due_date
- Aggiungere tempo stimato residuo
- Aggiungere stato corrente
- Rendere esplicito ciclo di vita ordine
- Eliminare dipendenze implicite tra strutture

---

## 4. Priorità EDD Dinamica Locale

- Ricalcolo tempo stimato quando macchina seleziona ordine
- Ordinamento coda secondo EDD aggiornato
- Nessun cambio routing
- Nessuna logica Digital Twin
- Validazione coerenza tempo simulato / stima

---

## 5. Arrivi Dinamici nel Mese

- 100 ordini iniziali
- 50–80 ordini distribuiti nel periodo
- Inserimento dinamico nel sistema
- Confronto performance con baseline

---

## 6. Digital Twin

- Stessi input e regole strutturali del modello base
- Attivazione cambi route locali
- Attivazione cambi priorità locali
- Intervento in caso di saturazioni critiche
- Confronto Base vs Digital Twin
- Valutazione resilienza e capacità adattiva

---

## 7. Evoluzione Output e Valutazione

- Confronto CSV completi vs cut320
- Definizione metriche di resilienza
- Analisi backlog a 320 ore
- Analisi saturazione macchine
- Analisi lateness sotto domanda variabile
